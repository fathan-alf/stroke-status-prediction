# Stroke Status Prediction

## Project Description
This project aims to predict whether a patient is at risk of having a stroke
based on health and demographic data. The prediction is made using the
**K-Nearest Neighbors (KNN)** algorithm, with imbalanced data handled using
**SMOTE (Synthetic Minority Oversampling Technique)**.

## Dataset
- **Source**: [Kaggle - Stroke Prediction Dataset](https://www.kaggle.com/datasets/fedesoriano/stroke-prediction-dataset?datasetId=1120859)
- **Size**: 5,110 rows, 12 columns
- **Target**: `stroke` column (0 = No Stroke, 1 = Stroke)

## Features Used
| Feature | Description |
|---|---|
| age | Patient's age |
| gender | Patient's gender |
| hypertension | History of hypertension (0/1) |
| heart_disease | History of heart disease (0/1) |
| ever_married | Marital status |
| work_type | Type of occupation |
| Residence_type | Type of residence area |
| avg_glucose_level | Average blood glucose level |
| bmi | Body Mass Index |
| smoking_status | Smoking status |

## 🔧 Data Preparation
The preprocessing steps include handling missing values in the `bmi` column
using the mean value, as well as binning the `age`, `avg_glucose_level`,
and `bmi` features to make them easier to analyze.

```python
# menangani missing values pada feature 'bmi' menggunakan mean/rata-rata
df['bmi'] = df['bmi'].fillna(round(df['bmi'].mean(), 1))
df.isna().sum()
```

```python
# binning data

# binning data untuk feature age
df['age_cat'] = pd.cut(df['age'], bins = [0,13,18,40,60,200], labels = ['Anak-anak', 'Remaja', 'Dewasa Muda', 'Dewasa Tua', 'Lanjut Usia'])

# binning data untuk glucose
df['glucose_cat'] = pd.cut(df['avg_glucose_level'], bins = [0,70,140,200,1000], labels = ['Rendah', 'Normal', 'Tinggi', 'Sangat Tinggi'])

# binning data untuk bmi
df['bmi_cat'] = pd.cut(df['bmi'], bins = [0,19,25,30,10000], labels = ['Kekurangan', 'Ideal', 'Kelebihan', 'Obesitas'])

df
```

## 🤖 Modelling — K-Nearest Neighbors (KNN)
The first model is built using KNN, with the best value of `k` determined
through Cross Validation (k=1 to k=20). The data is split into 70% training
and 30% testing.

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
# Encoding data
pred_data = df
label_encoder = LabelEncoder()
categorical_cols = ['gender', 'ever_married', 'work_type', 'Residence_type', 'smoking_status', 'age_cat', 'glucose_cat', 'bmi_cat']

for col in categorical_cols:
  pred_data[col] = label_encoder.fit_transform(pred_data[col])

pred_data
```

```python
x = df.drop(['stroke'], axis=1) # Penghapusan kolom target(stroke)
y = df['stroke'] # Kolom target(stroke)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=100)

print(f'Shape of training data : {x_train.shape}, {y_train.shape}')
print(f'Shape of testing data : {x_test.shape}, {y_test.shape}')
```

```python
# Menggunakan Cross Validation untuk mencari n terbaik untuk KNN(K-Nearest Neighbors)
neighbors = list(range(1, 21)) # mencari n terbaik mulai dari 1-20
cv_scores = []
for k in neighbors:
  knn = KNeighborsClassifier(n_neighbors=k)
  scores = cross_val_score(knn, x_train, y_train, cv=5, scoring='accuracy')
  cv_scores.append(scores.mean())
```

```python
# Mencari best n pada KNN
best_k = neighbors[cv_scores.index(max(cv_scores))]
print(f'Best n_neighbors: {best_k}')
```

```python
# Training data
knn = KNeighborsClassifier(n_neighbors=best_k)
knn.fit(x_train, y_train) # Perintah untuk melatih model KNN dengan data train
```

```python
# Testing data
knn_pred = knn.predict(x_test)
knn_conf = confusion_matrix(y_test, knn_pred)
knn_report = classification_report(y_test, knn_pred)
```

```python
# Show classification report
knn_acc = round(accuracy_score(y_test, knn_pred)*100, ndigits=2)
print(f'Confusion Matrix : \n\n{knn_conf}')
print(f'\nClassification Report : \n\n{knn_report}')
print(f'\nThe Accuracy of K Nearest Neighbors Classifier is {knn_acc}%')
```

## ⚖️ Oversampling with SMOTE
This dataset is highly imbalanced — only about 5% of patients are
stroke-positive. To address this, SMOTE is applied to synthesize new samples
for the minority class, allowing the model to learn more effectively.

```python
!pip install imbalanced-learn
```

```python
from imblearn.over_sampling import SMOTE
```

```python
# Membuat variable prediktor dan target
x = df.drop(['stroke', 'id'], axis=1) # Prediktor
y = df['stroke'] # Target

print(f'Shape of x before oversampling : {x.shape}')
print(f'Shape of y before oversampling : {y.shape}')
```

```python
# Memanggil SMOTE
smote = SMOTE()
```

```python
# Implementasi SMOTE pada variable prediktor dan target
x_resampled, y_resampled = smote.fit_resample(x, y)
print(f'Shape of x after oversampling : {x_resampled.shape}')
print(f'Shape of y after oversampling : {y_resampled.shape}')
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_resampled, y_resampled, test_size=0.3, random_state=100)
print(f'Shape of training data : {x_train.shape}, {y_train.shape}')
print(f'Shape of testing data : {x_test.shape}, {y_train.shape}')
```

```python
neighbors = list(range(1, 21))
cv_scores = []
for k in neighbors:
  knn = KNeighborsClassifier(n_neighbors=k)
  scores = cross_val_score(knn, x_train, y_train, cv=5, scoring='accuracy')
  cv_scores.append(scores.mean())
```

```python
# Mencari best n pada KNN
best_k = neighbors[cv_scores.index(max(cv_scores))]
print(f'Best n_neighbors: {best_k}')
```

```python
knn = KNeighborsClassifier(n_neighbors=best_k)
knn.fit(x_train, y_train)
```

```python
knn_pred = knn.predict(x_test)
knn_conf = confusion_matrix(y_test, knn_pred)
knn_report = classification_report(y_test, knn_pred)
knn_acc = round(accuracy_score(y_test, knn_pred)*100, ndigits=2)

print(f'Confusion Matrix : \n\n{knn_conf}')
print(f'\nClassification Report : \n\n{knn_report}')
print(f'\nThe Accuracy of K Nearest Neighbors Classifier is : {knn_acc}%')
```

## ✅ Conclusion
- The KNN model **before** SMOTE tends to be biased toward the majority
  class (no stroke)
- After applying SMOTE, the model's performance on the minority class (stroke)
  improved significantly, reflected in higher recall and F1-score values
- The most influential factors associated with stroke risk include:
  **older age, hypertension, heart disease, and high blood glucose levels**
